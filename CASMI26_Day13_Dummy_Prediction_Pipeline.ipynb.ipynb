```python
def mrr_at_25(predictions, true_molecules):
    scores = []

    for preds, true_molecule in zip(predictions, true_molecules):
        score = 0

        for rank, candidate in enumerate(preds[:25], start=1):
            if candidate == true_molecule:
                score = 1 / rank
                break

        scores.append(score)

    return sum(scores) / len(scores)
```

```python
molecules = (
    unique_molecules
    .drop_duplicates("inchikey14")
    .reset_index(drop=True)
)

print(molecules.shape)
```

```python
from rdkit.Chem import rdFingerprintGenerator

generator = rdFingerprintGenerator.GetMorganGenerator(
    radius=2,
    fpSize=2048
)

molecules["fingerprint"] = molecules["normalized_smiles"].apply(
    lambda x: generator.GetFingerprint(Chem.MolFromSmiles(x))
)

print(molecules.shape)
```

```python
import numpy as np
from sklearn.model_selection import train_test_split

molecule_ids = np.array(molecules["inchikey14"].tolist())

train_ids, val_ids = train_test_split(
    molecule_ids,
    test_size=0.20,
    random_state=42
)

print(len(train_ids))
print(len(val_ids))
```

```python
train_molecules = molecules[molecules["inchikey14"].isin(train_ids)]
val_molecules = molecules[molecules["inchikey14"].isin(val_ids)]

print(train_molecules.shape)
print(val_molecules.shape)
```

```python
from rdkit import DataStructs

val_fp = val_molecules["fingerprint"].iloc[0]

similarities = DataStructs.BulkTanimotoSimilarity(
    val_fp,
    train_molecules["fingerprint"].tolist()
)

print(max(similarities))
```

```python
sample_val = val_molecules.head(100)

max_similarities = []

for fp in sample_val["fingerprint"]:
    similarities = DataStructs.BulkTanimotoSimilarity(
        fp,
        train_molecules["fingerprint"].tolist()
    )
    max_similarities.append(max(similarities))

print(max_similarities[:10])
```

```python
import numpy as np

print("Minimum:", min(max_similarities))
print("Median:", np.median(max_similarities))
print("Maximum:", max(max_similarities))
```

```python
sample_val = val_molecules.head(1000)

max_similarities = []

for fp in sample_val["fingerprint"]:
    similarities = DataStructs.BulkTanimotoSimilarity(
        fp,
        train_molecules["fingerprint"].tolist()
    )
    max_similarities.append(max(similarities))

print("Minimum:", min(max_similarities))
print("Median:", np.median(max_similarities))
print("Maximum:", max(max_similarities))
```

```python
import numpy as np

threshold_low = np.percentile(max_similarities, 20)
threshold_high = np.percentile(max_similarities, 80)

print("20th percentile:", threshold_low)
print("80th percentile:", threshold_high)
```

```python
import pyarrow.parquet as pq

parquet_path = r"E:\project\Kaggle\data\train.parquet"

pf = pq.ParquetFile(parquet_path)

print("Number of row groups:", pf.num_row_groups)
print("Rows in first row group:", pf.metadata.row_group(0).num_rows)
```

```python
train_sample = pf.read_row_group(
    0,
    columns=[
        "inchikey14",
        "normalized_smiles",
        "ms2_mzs",
        "ms2_normalized_intensities"
    ]
).to_pandas()

print(train_sample.shape)
```

```python
query = train_sample.iloc[0]

print("True molecule:", query["inchikey14"])
print("Number of peaks:", len(query["ms2_mzs"]))
```

```python
candidates = train_sample["inchikey14"].drop_duplicates().sample(
    n=25,
    random_state=42
).tolist()

print(len(candidates))
print(candidates)
```

```python
import random

candidate_scores = {
    candidate: random.random()
    for candidate in candidates
}

ranked_candidates = sorted(
    candidate_scores,
    key=candidate_scores.get,
    reverse=True
)

print(ranked_candidates)
```

```python
true_molecule = query["inchikey14"]

if true_molecule in ranked_candidates:
    rank = ranked_candidates.index(true_molecule) + 1
else:
    rank = None

print("True molecule:", true_molecule)
print("Rank:", rank)
```

```python
score = mrr_at_25(
    [ranked_candidates],
    [true_molecule]
)

print("MRR@25:", score)
```
