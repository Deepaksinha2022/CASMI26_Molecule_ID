```python
def mrr_at_25(predictions, true_molecules):
    scores = []

    for preds, true_molecule in zip(predictions, true_molecules):
        score = 0

        for rank, candidate in enumerate(preds[:25], start=1):
            if candidate == true_molecule:
                score = 1 / rank
                break

        scores.append(score)

    return sum(scores) / len(scores)
```

```python
import numpy as np

def clean_spectrum(mzs, intensities):
    mzs = np.asarray(mzs, dtype=float)
    intensities = np.asarray(intensities, dtype=float)

    # Keep finite and positive peaks
    mask = (
        np.isfinite(mzs) &
        np.isfinite(intensities) &
        (intensities > 0)
    )

    mzs = mzs[mask]
    intensities = intensities[mask]

    # Sort by m/z
    order = np.argsort(mzs)

    return mzs[order], intensities[order]
```

```python
clean_mzs, clean_intensities = clean_spectrum(mzs, intensities)

print("Original peaks:", len(mzs))
print("Cleaned peaks:", len(clean_mzs))
print("First 10 cleaned m/z:", clean_mzs[:10])
```

```python
import numpy as np

bin_width = 1.0

min_mz = min(mzs_A.min(), mzs_B.min())
max_mz = max(mzs_A.max(), mzs_B.max())

bins = np.arange(
    np.floor(min_mz),
    np.ceil(max_mz) + bin_width,
    bin_width
)

print("Number of bins:", len(bins) - 1)
print("First 10 bins:", bins[:10])
```

```python
A_binned = np.zeros(len(bins) - 1)

bin_indices_A = np.digitize(mzs_A, bins) - 1

for idx, intensity in zip(bin_indices_A, int_A):
    A_binned[idx] += intensity

print("Non-zero bins in A:", np.count_nonzero(A_binned))
```

```python
B_binned = np.zeros(len(bins) - 1)

bin_indices_B = np.digitize(mzs_B, bins) - 1

for idx, intensity in zip(bin_indices_B, int_B):
    B_binned[idx] += intensity

print("Non-zero bins in B:", np.count_nonzero(B_binned))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

original_similarity = cosine_similarity(
    A_binned.reshape(1, -1),
    B_binned.reshape(1, -1)
)[0, 0]

print("Original cosine similarity:", original_similarity)
```

```python
A_sqrt = np.sqrt(A_binned)
B_sqrt = np.sqrt(B_binned)

sqrt_similarity = cosine_similarity(
    A_sqrt.reshape(1, -1),
    B_sqrt.reshape(1, -1)
)[0, 0]

print("Sqrt cosine similarity:", sqrt_similarity)
```

```python
def count_one_to_one_matches(mzs_A, mzs_B, tolerance=0.01):
    used_B = set()
    matches = 0

    for mz_A in mzs_A:
        distances = np.abs(mzs_B - mz_A)
        nearest = np.argmin(distances)

        if distances[nearest] <= tolerance and nearest not in used_B:
            matches += 1
            used_B.add(nearest)

    return matches
```

```python
matches_1to1 = count_one_to_one_matches(mzs_A, mzs_B)

print("One-to-one matches:", matches_1to1)
```

```python
similarity_1to1 = matches_1to1 / min(len(mzs_A), len(mzs_B))

print("One-to-one similarity:", similarity_1to1)
```

```python
def weighted_one_to_one_similarity(mzs_A, intensities_A, mzs_B, tolerance=0.01):
    used_B = set()
    matched_intensity = 0.0

    for mz_A, intensity_A in zip(mzs_A, intensities_A):
        distances = np.abs(mzs_B - mz_A)
        nearest = np.argmin(distances)

        if distances[nearest] <= tolerance and nearest not in used_B:
            matched_intensity += intensity_A
            used_B.add(nearest)

    return matched_intensity / np.sum(intensities_A)
```

```python
def retrieve_top25(query, candidate_pool, bins):
    query_mzs = query["ms2_mzs"]
    query_intensities = query["ms2_normalized_intensities"]

    # Create query vector
    query_vector = np.zeros(len(bins) - 1)

    indices = np.digitize(query_mzs, bins) - 1
    indices = np.clip(indices, 0, len(query_vector) - 1)

    for idx, intensity in zip(indices, query_intensities):
        query_vector[idx] += intensity

    similarities = []

    # Compare against candidates
    for _, candidate in candidate_pool.iterrows():
        vector = np.zeros(len(bins) - 1)

        indices = np.digitize(candidate["ms2_mzs"], bins) - 1
        indices = np.clip(indices, 0, len(vector) - 1)

        for idx, intensity in zip(
            indices,
            candidate["ms2_normalized_intensities"]
        ):
            vector[idx] += intensity

        score = cosine_similarity(
            query_vector.reshape(1, -1),
            vector.reshape(1, -1)
        )[0, 0]

        similarities.append(score)

    # Top 25 spectra
    top_indices = np.argsort(similarities)[::-1][:25]

    retrieved = candidate_pool.iloc[top_indices].copy()
    retrieved["similarity"] = np.array(similarities)[top_indices]

    # Remove query itself
    retrieved = retrieved[
        retrieved.index != query.name
    ]

    # Molecule-level deduplication
    retrieved = (
        retrieved
        .sort_values("similarity", ascending=False)
        .drop_duplicates("inchikey14")
    )

    return retrieved.head(25)
```
