```python
def mrr_at_25(predictions, true_molecules):
    scores = []

    for preds, true_molecule in zip(predictions, true_molecules):
        score = 0

        for rank, candidate in enumerate(preds[:25], start=1):
            if candidate == true_molecule:
                score = 1 / rank
                break

        scores.append(score)

    return sum(scores) / len(scores)
```

```python
predictions = [
    ["A", "B", "C", "D"],
    ["X", "Y", "Z"],
    ["P", "Q", "R"]
]

true_molecules = ["A", "Z", "M"]

mrr_at_25(predictions, true_molecules)
```

```python
predictions = [
    ["A"],                    # rank 1
    ["X", "A"],               # rank 2
    ["X", "Y", "A"],          # rank 3
    ["X", "Y", "Z"],          # absent
]

true_molecules = ["A", "A", "A", "A"]

mrr_at_25(predictions, true_molecules)
```

```python
test_predictions = [
    ["A", "B", "C"],
    ["X", "A", "Y"],
    ["P", "Q", "R"]
]

test_true = ["A", "A", "Z"]

test_score = mrr_at_25(test_predictions, test_true)

print("Test MRR@25:", test_score)
```
